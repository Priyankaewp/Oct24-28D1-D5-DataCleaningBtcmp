```python
# handling missing age
# there are outliers, hence we use median
df["age"] = df["age"].fillna(df["age"].median())
print(df['age'].isnull().sum())
```

```python
# handling missing embarked
#  embarked is a categorical value, hence we use = mode()
df["embarked"] = df["embarked"].fillna(df["embarked"].mode())
print(df['embarked'].isnull().sum())

df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
print(df['embarked'].isnull().sum())
```

```python
# handling missing embark_town data
#  embark_town is a categorical value, hence we use = mode()
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode())
print(df['embark_town'].isnull().sum())
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
print(df['embark_town'].isnull().sum())
```

```python
df["deck"] = df["deck"].fillna("unknown")
print(df['deck'].isnull().sum())
```

```python
print(df.isnull().sum())
```

```python
# duplicate records
print("Duplicate records:" , df.duplicated().sum())
df = df.drop_duplicates()
print('duplicates removed!')
```
