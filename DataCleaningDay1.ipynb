```python
# Load the excel file into collab
# import the required libraries
import numpy as np
import pandas as pd
```

```python
#  load the data from the excel file - day1.xlsx
df = pd.read_excel("day1.xlsx")
```

```python
# check how the original dataset looks like
print(df)
```

```python
# inspect the data
# check the rows and column's data
# Rows = 0, Columns = 1

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
# Find the missing values
print(df.isnull().sum())
```

```python
# check for the rows that contain missing values
# isnull() - checks for the missing values
# any(axis=1) gives the rocord or the rows where isnull is true or any row that has a missing value(s).

print(df[df.isnull().any(axis=1)])
```

```python
# check for duplicate values
print(df.duplicated().sum())
```

```python
# check how many duplicate customer ids are present.
print(df['CustomerID'].duplicated().sum())
```

```python
# Check Unique Values
print(df.nunique())
```

```python
# check for the outlier or the out of range values.
#  eg: age where age is -3 and 150
suspicious = df[(df['Age']<0) | (df['Age']> 100)]
print(suspicious)
```
